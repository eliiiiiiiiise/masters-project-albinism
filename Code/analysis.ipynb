## 2- reading the current datafile

```python
current_data = pd.read_csv("current_dataset.csv")
current_full_data = pd.read_csv("current_full_dataset_v2.csv")
```

```python
# making a copy of the dataframe
current_data_copy = current_data.copy()
current_full_data_copy = current_full_data.copy()
```

```python
# changing the column types of theme and country
current_data_copy["theme"] = current_data_copy["theme"].astype(object)
current_data_copy["country"] = current_data_copy["country"].astype(object)
```

```python
# getting the data that has not been analysed yet
to_merge_data = current_data_copy[["url", "title", "authors", "date", "extract"]]
to_merge_full_data = current_full_data_copy[["url", "title", "authors", "date", "extract"]]
merged_data = to_merge_data.merge(to_merge_full_data, how='left', indicator=True)
```

```python
data_to_analyse =  current_data_copy[merged_data["_merge"] == "left_only"]
data_to_analyse
```

## 3- loading the places that have already been mapped

```python
with open("place_country.json", "r", encoding="utf-8") as f:
    map_dic = json.load(f)
```

## 4- doing the analysis

```python
new_full_data, new_map_dic = analyse(data_to_analyse, map_dic=map_dic)
```

## 5- saving the dataset with the analyses

```python
current_full_data = pd.concat([current_full_data_copy, new_full_data], ignore_index=True)
```

```python
current_full_data.to_csv("current_full_dataset_v2.csv", index=False)
```
